```python
from google.cloud import storage 

# Configuración
PROJECT_ID = "inventarioproject-471611"
BUCKET_NAME = "proyecto_inventario_bucket"

# Ruta de tu archivo JSON
credenciales_path = "F:\Descargas\inventarioproject-471611-6b87dc862c06.json"



# Crear cliente con credenciales explícitas
client = storage.Client.from_service_account_json(credenciales_path)

try:
    bucket = client.bucket(BUCKET_NAME)
    print(f"✅ Conexión exitosa al bucket: {BUCKET_NAME}")

    # Listar archivos
    blobs = list(bucket.list_blobs())
    if not blobs:
        print("⚠️ El bucket está vacío.")
    else:
        print("📂 Archivos encontrados en el bucket:")
        for blob in blobs:
            print("-", blob.name)
except Exception as e:
    print("❌ Error al conectar o listar archivos:")
    print(e)

```

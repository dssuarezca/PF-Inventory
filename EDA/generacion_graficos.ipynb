### importamos la librerias que vamos a necesitar

```python
#en primer instancia importamos las librerias necesarias para poder trabajar con los datos, en este caso utilizamos pandas para poder leer el archivo csv 
# y poder realizar diferentes operaciones al dataset
import pandas as pd
import numpy as np
import seaborn as sns
import matplotlib.pyplot as plt
import matplotlib.colors as mcolors   
```

```python
#leemos el archivo csv y lo almacenamos en un dataframe
df_inventario_inicial = pd.read_csv("F:\Documentos\PF Inventory\BegInvFINAL12312016.csv")
df_inventario_inicial["startDate"] = pd.to_datetime(df_inventario_inicial["startDate"])
```

```python
#leemos el archivo csv y lo almacenamos en un dataframe
df_inventario_final = pd.read_csv("F:\Documentos\PF Inventory\data clean\Inventario_final_2016_clean.csv")
```

### generamos los gráficos del inventario inicial

```python

# Seleccionar los 20 productos de mayor valor
top20 = df_inventario_inicial.sort_values("Price", ascending=False).head(20)

# Graficar
plt.figure(figsize=(12,6))
sns.barplot(data=top20, x="Price", y="Description", palette="viridis")  

plt.title("Top 20 productos con mayor valor", fontsize=14)
plt.xlabel("Precio")
plt.ylabel("Producto")
plt.show()
```

```python
# Agrupamos por productos para saber el stock total por producto 
productos_top = (
    df_inventario_inicial.groupby("Description")["onHand"]
      .sum()
      .sort_values(ascending=False)
      .head(20)
      .reset_index()
)

# Gráfico
plt.figure(figsize=(10,6))
sns.barplot(
    data=productos_top,
    x="onHand",
    y="Description",
    palette="viridis"
)
plt.title("Top 20 Productos con mayor stock")
plt.xlabel("Stock")
plt.ylabel("Description")
plt.show()
```

```python
# Agrupamos por ciudades para saber el stock total por ciudad 
productos_top = (
    df_inventario_inicial.groupby("City")["onHand"]
      .sum()
      .sort_values(ascending=False)
      .head(20)
      .reset_index()
)

# Gráfico
plt.figure(figsize=(10,6))
sns.barplot(
    data=productos_top,
    x="onHand",
    y="City",
    palette="viridis"
)
plt.title("Top 20 ciudades con mayor stock")
plt.xlabel("Stock")
plt.ylabel("City")
plt.show()
```

```python
df_inventario_inicial.columns = ["ID Inventario", "Tienda", "Ciudad", "Marca", "Descripción", "Volumen", "Stock", "Precio", "Fecha Inventario"]
```

### generamos los gráficos del inventario final

```python

# Seleccionar los 20 productos de mayor valor
top20 = df_inventario_final.sort_values("Precio Unitario", ascending=False).head(20)

# Graficar
plt.figure(figsize=(12,6))
sns.barplot(data=top20, x="Precio Unitario", y="Descripción", palette="viridis")

plt.title("Top 20 productos con mayor valor", fontsize=14)
plt.xlabel("Precio")
plt.ylabel("Producto")
plt.show()
```

```python
# Agrupamos por productos para saber el stock total por producto 
productos_top = (
    df_inventario_final.groupby("Descripción")["Stock"]
      .sum()
      .sort_values(ascending=False)
      .head(20)
      .reset_index()
)

# Gráfico
plt.figure(figsize=(10,6))
sns.barplot(
    data=productos_top,
    x="Stock",
    y="Descripción",
    palette="viridis"
)
plt.title("Top 20 Productos con mayor stock")
plt.xlabel("Stock")
plt.ylabel("Descripción")
plt.show()
```

```python
# Agrupamos por productos para saber el stock total por producto 
productos_top = (
    df_inventario_final.groupby("Ciudad")["Stock"]
      .sum()
      .sort_values(ascending=False)
      .head(20)
      .reset_index()
)

# Gráfico
plt.figure(figsize=(10,6))
sns.barplot(
    data=productos_top,
    x="Stock",
    y="Ciudad",
    palette="viridis"
)
plt.title("Top 20 ciudades con mayor stock")
plt.xlabel("Stock")
plt.ylabel("Ciudad")
plt.show()
```

### unimos los dos graficos para saber que producto tuvo mayor demanda 

```python
# Unir ambas tablas por producto
df_inventariogeneral = pd.merge(df_inventario_inicial, df_inventario_final, on=["Descripción", "Tienda"])
```

### calculamos el producto mas demandado  

```python
#calculamos la demanda
df_inventariogeneral["demanda"] = df_inventariogeneral["Stock_x"] - df_inventariogeneral["Stock_y"]

# Agrupar por producto y sumar la demanda
demanda_por_producto = df_inventariogeneral.groupby("Descripción")["demanda"].sum().reset_index()

# Ordenar de mayor a menor demanda
demanda_por_producto = demanda_por_producto.sort_values(by="demanda", ascending=False)

plt.figure(figsize=(10,6))
sns.barplot(data=demanda_por_producto.head(20), 
            x="demanda", 
            y="Descripción", 
            palette="viridis")

plt.title("Top 10 productos con mayor demanda", fontsize=14)
plt.xlabel("Demanda total")
plt.ylabel("Producto")
plt.show()
```

### calculamos las tiendas mas demandadas

```python

# Agrupar por producto y sumar la demanda
demanda_por_tienda = df_inventariogeneral.groupby("Ciudad_x")["demanda"].sum().reset_index()

# Ordenar de mayor a menor demanda
demanda_por_tienda = demanda_por_tienda.sort_values(by="demanda", ascending=False)

plt.figure(figsize=(10,6))
sns.barplot(data=demanda_por_tienda.head(20), 
            x="demanda", 
            y="Ciudad_x", 
            palette="Blues_r")

plt.title("Top 10 tiendas con mayor demanda", fontsize=14)
plt.xlabel("Demanda total")
plt.ylabel("Ciudad")
plt.show()
```

```python
#leemos el archivo csv y lo almacenamos en un dataframe
df_factura_compras = pd.read_csv("F:\Documentos\PF Inventory\data clean\ComprasconFactura_2016_clean.csv")
```

```python
# Contar cuántos aprobados y no aprobados 
conteo = df_factura_compras["Estado"].value_counts()

# Crear gráfico de barras
plt.figure(figsize=(6,4))
conteo.plot(kind="bar", color=["#1417BEDA", "#13A1B4C6"])
plt.title("Productos aprobados vs no aprobados")
plt.xlabel("Estado de aprobación")
plt.ylabel("Cantidad de productos")
plt.xticks(rotation=0)
plt.show()
```

```python
# 2. Agrupar por proveedor y estado de aprobación, sumando el total
grouped = df_factura_compras.groupby(["Vendedor", "Estado"])["Total"].sum().reset_index()

# 3. Pivotear para tener columnas separadas (Aprobado vs No aprobado)
pivot = grouped.pivot(index="Vendedor", columns="Estado", values="Total").fillna(0)

# 4. Ordenar por el total mayor (suma de ambas columnas) y quedarnos con los top 10
pivot["TotalGeneral"] = pivot.sum(axis=1)
pivot_top10 = pivot.sort_values("TotalGeneral", ascending=False).head(10)

# 5. Graficar columnas agrupadas
pivot_top10.drop(columns="TotalGeneral").plot(
    kind="bar", 
    color=["#1417BEDA", "#13A1B4C6"],
    figsize=(12,6)
)

plt.title("Top 10 Proveedores en compras")
plt.ylabel("$ Total")
plt.xlabel("Proveedor")
plt.xticks(rotation=45, ha="right")
plt.legend(title="Estado")
plt.tight_layout()
plt.show()
```

```python
# 2. Agrupar por proveedor y estado de aprobación, sumando el total
grouped = df_factura_compras.groupby(["Vendedor", "Estado"])["Cantidad"].sum().reset_index()

# 3. Pivotear para tener columnas separadas (Aprobado vs No aprobado)
pivot = grouped.pivot(index="Vendedor", columns="Estado", values="Cantidad").fillna(0)

# 4. Ordenar por el total mayor (suma de ambas columnas) y quedarnos con los top 10
pivot["TotalGeneral"] = pivot.sum(axis=1)
pivot_top10 = pivot.sort_values("TotalGeneral", ascending=False).head(10)

# 5. Graficar columnas agrupadas
pivot_top10.drop(columns="TotalGeneral").plot(
    kind="bar", 
    color=["#1417BEDA", "#13A1B4C6"],
    figsize=(12,6)
)

plt.title("Top 10 Proveedores compras en cantidad")
plt.ylabel("Cantidad")
plt.xlabel("Proveedor")
plt.xticks(rotation=45, ha="right")
plt.legend(title="Estado")
plt.tight_layout()
plt.show()
```

```python
# Agrupamos por cantidad de productos para saber cual es el pedido mas vendido 
productos_top = (
    df_ventas_finales.groupby("description")["salesquantity"]
      .sum()
      .sort_values(ascending=False)
      .head(15)
      .reset_index()
)

# Gráfico
plt.figure(figsize=(10,6))
sns.barplot(
    data=productos_top,
    x="description",
    y="salesquantity",
    palette="viridis"
)
plt.title("Top 20 ciudades con mayor cantidad solicitada")
plt.xlabel("Producto")
plt.ylabel("Cantidad")
plt.xticks(rotation=45, ha="right")
plt.show()
```

```python
#calculamos el total de venta por producto
df_ventas_finales["Total_venta"] = df_ventas_finales["excisetax"]*df_ventas_finales["salesquantity"]+df_ventas_finales["salesdollars"]
```

```python
# Agrupamos por cantidad de productos para saber cual es el pedido mas vendido 
productos_top = (
    df_ventas_finales.groupby("description")["Total_venta"]
      .sum()
      .sort_values(ascending=False)
      .head(15)
      .reset_index()
)

# Gráfico
plt.figure(figsize=(10,6))
sns.barplot(
    data=productos_top,
    x="description",
    y="Total_venta",
    palette="viridis"
)
plt.title("Top 20 productos mas costosos")
plt.xlabel("Producto")
plt.ylabel("Valor Total")
plt.xticks(rotation=45, ha="right")
plt.show()
```

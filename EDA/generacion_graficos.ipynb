### importamos la librerias que vamos a necesitar

```python
#en primer instancia importamos las librerias necesarias para poder trabajar con los datos, en este caso utilizamos pandas para poder leer el archivo csv 
# y poder realizar diferentes operaciones al dataset
import pandas as pd
import numpy as np
import seaborn as sns
import matplotlib.pyplot as plt
```

```python
#leemos el archivo csv y lo almacenamos en un dataframe
df_inventario_inicial = pd.read_csv("F:\Documentos\PF Inventory\BegInvFINAL12312016.csv")
df_inventario_inicial["startDate"] = pd.to_datetime(df_inventario_inicial["startDate"])
```

```python
#leemos el archivo csv y lo almacenamos en un dataframe
df_inventario_final = pd.read_csv("F:\Documentos\PF Inventory\data clean\Inventario_final_2016_clean.csv")
```

### generamos los gráficos del inventario inicial

```python

# Seleccionar los 20 productos de mayor valor
top20 = df_inventario_inicial.sort_values("Price", ascending=False).head(20)

# Graficar
plt.figure(figsize=(12,6))
sns.barplot(data=top20, x="Price", y="Description", palette="viridis")  

plt.title("Top 20 productos con mayor valor", fontsize=14)
plt.xlabel("Precio")
plt.ylabel("Producto")
plt.show()
```

```python
# Agrupamos por productos para saber el stock total por producto 
productos_top = (
    df_inventario_inicial.groupby("Description")["onHand"]
      .sum()
      .sort_values(ascending=False)
      .head(20)
      .reset_index()
)

# Gráfico
plt.figure(figsize=(10,6))
sns.barplot(
    data=productos_top,
    x="onHand",
    y="Description",
    palette="viridis"
)
plt.title("Top 20 Productos con mayor stock")
plt.xlabel("Stock")
plt.ylabel("Description")
plt.show()
```

```python
# Agrupamos por ciudades para saber el stock total por ciudad 
productos_top = (
    df_inventario_inicial.groupby("City")["onHand"]
      .sum()
      .sort_values(ascending=False)
      .head(20)
      .reset_index()
)

# Gráfico
plt.figure(figsize=(10,6))
sns.barplot(
    data=productos_top,
    x="onHand",
    y="City",
    palette="viridis"
)
plt.title("Top 20 ciudades con mayor stock")
plt.xlabel("Stock")
plt.ylabel("City")
plt.show()
```

```python
df_inventario_inicial.columns = ["ID Inventario", "Tienda", "Ciudad", "Marca", "Descripción", "Volumen", "Stock", "Precio", "Fecha Inventario"]
```

### generamos los gráficos del inventario final

```python

# Seleccionar los 20 productos de mayor valor
top20 = df_inventario_final.sort_values("Precio Unitario", ascending=False).head(20)

# Graficar
plt.figure(figsize=(12,6))
sns.barplot(data=top20, x="Precio Unitario", y="Descripción", palette="viridis")

plt.title("Top 20 productos con mayor valor", fontsize=14)
plt.xlabel("Precio")
plt.ylabel("Producto")
plt.show()
```

```python
# Agrupamos por productos para saber el stock total por producto 
productos_top = (
    df_inventario_final.groupby("Descripción")["Stock"]
      .sum()
      .sort_values(ascending=False)
      .head(20)
      .reset_index()
)

# Gráfico
plt.figure(figsize=(10,6))
sns.barplot(
    data=productos_top,
    x="Stock",
    y="Descripción",
    palette="viridis"
)
plt.title("Top 20 Productos con mayor stock")
plt.xlabel("Stock")
plt.ylabel("Descripción")
plt.show()
```

```python
# Agrupamos por productos para saber el stock total por producto 
productos_top = (
    df_inventario_final.groupby("Ciudad")["Stock"]
      .sum()
      .sort_values(ascending=False)
      .head(20)
      .reset_index()
)

# Gráfico
plt.figure(figsize=(10,6))
sns.barplot(
    data=productos_top,
    x="Stock",
    y="Ciudad",
    palette="viridis"
)
plt.title("Top 20 ciudades con mayor stock")
plt.xlabel("Stock")
plt.ylabel("Ciudad")
plt.show()
```

### unimos los dos graficos para saber que producto tuvo mayor demanda 

```python
# Unir ambas tablas por producto
df_inventariogeneral = pd.merge(df_inventario_inicial, df_inventario_final, on=["Descripción", "Tienda"])
```

### calculamos el producto mas demandado  

```python
#calculamos la demanda
df_inventariogeneral["demanda"] = df_inventariogeneral["Stock_x"] - df_inventariogeneral["Stock_y"]

# Agrupar por producto y sumar la demanda
demanda_por_producto = df_inventariogeneral.groupby("Descripción")["demanda"].sum().reset_index()

# Ordenar de mayor a menor demanda
demanda_por_producto = demanda_por_producto.sort_values(by="demanda", ascending=False)

plt.figure(figsize=(10,6))
sns.barplot(data=demanda_por_producto.head(20), 
            x="demanda", 
            y="Descripción", 
            palette="viridis")

plt.title("Top 10 productos con mayor demanda", fontsize=14)
plt.xlabel("Demanda total")
plt.ylabel("Producto")
plt.show()
```

### calculamos las tiendas mas demandadas

```python

# Agrupar por producto y sumar la demanda
demanda_por_tienda = df_inventariogeneral.groupby("Ciudad_x")["demanda"].sum().reset_index()

# Ordenar de mayor a menor demanda
demanda_por_tienda = demanda_por_tienda.sort_values(by="demanda", ascending=False)

plt.figure(figsize=(10,6))
sns.barplot(data=demanda_por_tienda.head(20), 
            x="demanda", 
            y="Ciudad_x", 
            palette="viridis")

plt.title("Top 10 tiendas con mayor demanda", fontsize=14)
plt.xlabel("Demanda total")
plt.ylabel("Ciudad")
plt.show()
```
